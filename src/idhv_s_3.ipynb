Cell 1: Load Data & Inspect Variables

```python
# ==============================================================================
# CELL 1: UPLOAD CLEANED_DATA.CSV & LOAD DATASET
# ==============================================================================
import pandas as pd
import numpy as np
import io
from google.colab import files

print("Please upload 'cleaned_data.csv':")
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("="*60)
print(f"DATASET LOADED: {df.shape[0]} Observations x {df.shape[1]} Attributes")
print("="*60)

# Identify numerical and categorical subsets (on native 1-5 scale)
numeric_cols = ['ad_trust', 'discount_importance', 'choose_promoted_product_likelihood', 'ad_receptivity_score']
categorical_cols = ['age_group', 'discovery_platform', 'ad_frequency', 'online_product_category',
                    'clicked_ad_3m', 'purchased_after_ad_3m', 'preferred_promotion', 'ad_ignore_reason']

print(f"✓ Quantitative Attributes ({len(numeric_cols)}): {numeric_cols}")
print(f"✓ Qualitative Attributes ({len(categorical_cols)}): {categorical_cols}")
```

Cell 2: Numerical Summary Statistics (Derivation & Interpretation)

```python
# ==============================================================================
# CELL 2: COMPREHENSIVE NUMERICAL DESCRIPTIVE STATISTICS
# ==============================================================================
# 1. Derive Central Tendency, Dispersion, and Shape metrics
stats_summary = pd.DataFrame(index=numeric_cols)

stats_summary['Mean'] = df[numeric_cols].mean()
stats_summary['Median'] = df[numeric_cols].median()
stats_summary['Mode'] = df[numeric_cols].mode().iloc[0]
stats_summary['Std_Dev'] = df[numeric_cols].std()
stats_summary['Variance'] = df[numeric_cols].var()
stats_summary['IQR'] = df[numeric_cols].quantile(0.75) - df[numeric_cols].quantile(0.25)
stats_summary['Min'] = df[numeric_cols].min()
stats_summary['Max'] = df[numeric_cols].max()
stats_summary['Skewness'] = df[numeric_cols].skew()
stats_summary['Kurtosis'] = df[numeric_cols].kurtosis()

print("="*85)
print("TABLE 1: DERIVED SUMMARY STATISTICS FOR NUMERICAL & LIKERT ATTRIBUTES")
print("="*85)
display(stats_summary.round(3))

# 2. Automated Statistical Interpretation Printout
print("\n" + "="*85)
print("INTERPRETATION OF SUMMARY STATISTICS:")
print("="*85)
for col in numeric_cols:
    mean_val = stats_summary.loc[col, 'Mean']
    median_val = stats_summary.loc[col, 'Median']
    skew_val = stats_summary.loc[col, 'Skewness']
    iqr_val = stats_summary.loc[col, 'IQR']

    skew_desc = "negatively skewed (left-tailed, leaning favorable)" if skew_val < -0.1 else \
                ("positively skewed (right-tailed, leaning conservative)" if skew_val > 0.1 else "approximately symmetric")

    print(f"• {col.upper()}:")
    print(f"   - Central Point : Mean = {mean_val:.2f}, Median = {median_val:.2f} (Center of scale: 3.00)")
    print(f"   - Distribution  : Skewness = {skew_val:.2f} -> {skew_desc}")
    print(f"   - Spread / Agree: IQR = {iqr_val:.2f}, Std Dev = {stats_summary.loc[col, 'Std_Dev']:.2f}")
```

Cell 3: Categorical Distributions (Frequency & Proportion Tables)

```python
# ==============================================================================
# CELL 3: CATEGORICAL FREQUENCY & PERCENTAGE SHARE DISTRIBUTIONS
# ==============================================================================
print("="*75)
print("TABLE 2: MARGINAL FREQUENCY & PROPORTION ANALYSIS")
print("="*75)

key_categories = ['discovery_platform', 'preferred_promotion', 'ad_ignore_reason', 'age_group']

for col in key_categories:
    counts = df[col].value_counts()
    shares = (df[col].value_counts(normalize=True) * 100).round(1)

    freq_table = pd.DataFrame({'Frequency (n)': counts, 'Share (%)': shares})
    freq_table['Cumulative (%)'] = freq_table['Share (%)'].cumsum().round(1)

    print(f"\n--- Distribution: {col.replace('_', ' ').title()} ---")
    display(freq_table)

    # Interpretation highlight
    top_cat = freq_table.index[0]
    top_share = freq_table.iloc[0]['Share (%)']
    print(f"  → Key Insight: '{top_cat}' is the dominant category, representing {top_share}% of total respondents.")
```

Cell 4: Marketing Conversion Funnel & Behavioral Ratios

```python
# ==============================================================================
# CELL 4: MARKETING CONVERSION FUNNEL & CROSS-FEATURE RATIOS
# ==============================================================================
total_n = len(df)
clicked_ad_count = (df['clicked_ad_3m'] == 'Yes').sum()
purchased_ad_count = (df['purchased_after_ad_3m'] == 'Yes').sum()

click_through_rate = (clicked_ad_count / total_n) * 100
purchase_conversion_rate = (purchased_ad_count / total_n) * 100
click_to_buy_efficiency = (purchased_ad_count / clicked_ad_count) * 100 if clicked_ad_count > 0 else 0

funnel_df = pd.DataFrame({
    'Funnel Stage': ['1. Total Audience', '2. Ad Clickers (Past 3 Months)', '3. Ad Purchasers (Direct Conversion)'],
    'Count': [total_n, clicked_ad_count, purchased_ad_count],
    'Stage Conversion Rate (%)': [100.0, click_through_rate, purchase_conversion_rate],
    'Step-to-Step Dropoff (%)': [0.0, 100.0 - click_through_rate, 100.0 - click_to_buy_efficiency]
})

print("="*80)
print("TABLE 3: DIGITAL ADVERTISING CONVERSION FUNNEL METRICS")
print("="*80)
display(funnel_df.round(2))

print("\nFUNNEL INTERPRETATION:")
print(f"• Click-Through Propensity: {click_through_rate:.1f}% of respondents actively click on ads.")
print(f"• Net Commercial Conversion: {purchase_conversion_rate:.1f}% of the total sample made a purchase.")
print(f"• Click-to-Purchase Efficiency: Once an ad is clicked, {click_to_buy_efficiency:.1f}% successfully convert.")
```

Cell 5: Bivariate Group Comparisons & Cross-Tabulations

```python
# ==============================================================================
# CELL 5: BIVARIATE SUB-GROUP ANALYSIS & CONTINGENCY TABLES
# ==============================================================================
# Ensure receptivity score exists (on native 1-5 scale)
if 'ad_receptivity_score' not in df.columns and 'ad_receptivity_index' not in df.columns:
    likert_cols = ['ad_trust', 'discount_importance', 'choose_promoted_product_likelihood']
    df['ad_receptivity_score'] = df[likert_cols].mean(axis=1).round(2)

score_col = 'ad_receptivity_score' if 'ad_receptivity_score' in df.columns else 'ad_receptivity_index'

print("="*85)
print("TABLE 4: CROSS-TABULATION — AGE GROUP VS. PREFERRED PROMOTION INCENTIVE")
print("="*85)

# Two-way contingency table
contingency_table = pd.crosstab(
    df['age_group'],
    df['preferred_promotion'],
    margins=True,
    margins_name="Total"
)
display(contingency_table)

print("\n" + "="*85)
print(f"TABLE 5: SUB-GROUP METRICS — AD RECEPTIVITY & TRUST BY DISCOVERY PLATFORM")
print("="*85)

# Groupby metrics across platforms using native 1-5 scale
platform_metrics = df.groupby('discovery_platform').agg(
    Sample_Size=('ad_trust', 'count'),
    Mean_Trust=('ad_trust', 'mean'),
    Median_Trust=('ad_trust', 'median'),
    Mean_Receptivity=(score_col, 'mean'),
    Std_Receptivity=(score_col, 'std')
).round(2).sort_values(by='Mean_Receptivity', ascending=False)

display(platform_metrics)

print("\nGROUP COMPARISON INTERPRETATION:")
highest_plat = platform_metrics.index[0]
print(f"• Highest Promotional Responsiveness is observed on '{highest_plat}' (Score: {platform_metrics.loc[highest_plat, 'Mean_Receptivity']:.2f} / 5.0).")
print("• Immediate price discounts lead across all demographic age brackets, followed by Free Delivery.")
```

Cell 6: Correlation Analysis (Pearson & Spearman Rank Associations)

```python
# ==============================================================================
# CELL 6: CORRELATION MATRIX (PEARSON & SPEARMAN RANK)
# ==============================================================================
score_col = 'ad_receptivity_score' if 'ad_receptivity_score' in df.columns else 'ad_receptivity_index'
numeric_cols = ['ad_trust', 'discount_importance', 'choose_promoted_product_likelihood', score_col]

# Pearson evaluates linear relationships; Spearman evaluates monotonic ordinal relationships
pearson_corr = df[numeric_cols].corr(method='pearson').round(3)
spearman_corr = df[numeric_cols].corr(method='spearman').round(3)

print("="*75)
print("TABLE 6: PEARSON CORRELATION MATRIX (LINEAR ASSOCIATIONS)")
print("="*75)
display(pearson_corr)

print("\n" + "="*75)
print("TABLE 7: SPEARMAN RANK CORRELATION MATRIX (MONOTONIC ASSOCIATIONS)")
print("="*75)
display(spearman_corr)

print("\nCORRELATION INTERPRETATION:")
r_trust_promo = pearson_corr.loc['ad_trust', 'choose_promoted_product_likelihood']
r_disc_promo = pearson_corr.loc['discount_importance', 'choose_promoted_product_likelihood']
print(f"• Ad Trust vs. Promotion Choice Likelihood: r = {r_trust_promo:.3f}")
print(f"• Discount Importance vs. Promotion Choice Likelihood: r = {r_disc_promo:.3f}")
print("• The composite 'ad_receptivity_score' demonstrates balanced correlation across all underlying survey pillars.")
```
